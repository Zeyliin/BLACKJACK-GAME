```python
import random       
```

```python
class Card:
    def __init__(self, rank: str, suit: str):
        self.rank = rank
        self.suit = suit

    @property
    def points(self) -> int:
        if self.rank in ("J", "Q", "K"):
            return 10
        if self.rank == "A":
            return 11
        return int(self.rank)

    def __str__(self) -> str:
        return f"{self.rank} {self.suit}"

```

```python
class Deck:
    def __init__(self):
        suits = ["Kupa", "Karo", "Sinek", "Maça"]
        ranks = [str(number) for number in range(2, 11)] + ["J", "Q", "K", "A"]

        self.cards = [
            Card(rank, suit)
            for suit in suits
            for rank in ranks
        ]
        random.shuffle(self.cards)

    def draw(self) -> Card:
        return self.cards.pop()
```

```python
class BlackjackGame:
    def __init__(self):
        self.score = 100
        self.lives = 3

    @staticmethod
    def hand_total(hand: list[Card]) -> int:
        total = sum(card.points for card in hand)
        aces = sum(card.rank == "A" for card in hand)

        while total > 21 and aces > 0:
            total -= 10
            aces -= 1

        return total
@staticmethod
    def show_hand(name: str, hand: list[Card]) -> None:
        cards = ", ".join(str(card) for card in hand)
        total = BlackjackGame.hand_total(hand)
        print(f"{name}: {cards} | Toplam: {total}")

    def finish_round(self, result: str) -> None:
        if result == "win":
            points = random.randint(10, 30)
            self.score += points
            print(f"Eli kazandın! +{points} puan. Puanın: {self.score}")

        elif result == "loss":
            points = random.randint(10, 30)
            self.score -= points
            print(f"Eli kaybettin! -{points} puan. Puanın: {self.score}")

            if self.score <= 0:
                self.lives -= 1

                if self.lives > 0:
                    self.score = 100
                    print(
                        f"Bir can kaybettin. Kalan can: {self.lives}. "
                        "Puanın 100 oldu."
                    )
                else:
                    print("Üç canını da kaybettin!")

        else:
            print("Berabere! Puanın değişmedi.")

```
